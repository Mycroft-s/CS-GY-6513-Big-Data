```python
from pymongo import MongoClient

# connect
client = MongoClient(
    host="mongo-csgy-6513-spring.db",
    port=27017,
    username="hm3424",
    password="bigdata",
    authSource="bigdata"
)
db = client.bigdata
coll = db.book_sales
```

```python
# a) top 5 highest-rated books
top5 = list(coll.find(
    {}, 
    {"_id":0, "title":1, "average_rating":1}
).sort("average_rating", -1).limit(5))
print("Top 5 highest-rated books:")
for book in top5:
    print(f"  {book['title']} — {book['average_rating']}")
```

```python
# b) number of books published after 2010
count_post2010 = coll.count_documents({"publication_year": {"$gt": 2010}})
print("\nBooks published after 2010:", count_post2010)
```

```python
# c) average rating of books with more than 100,000 ratings
agg = coll.aggregate([
    {"$match": {"ratings_count": {"$gt": 100000}}},
    {"$group": {"_id": None, "avgRating": {"$avg": "$average_rating"}}}
])
avg_rating = next(agg)["avgRating"]
print("\nAverage rating (ratings_count > 100k):", round(avg_rating, 2))
```

```python
# d) distinct languages
languages = coll.distinct("language_code")
print("\nDistinct language codes:", languages)
```
